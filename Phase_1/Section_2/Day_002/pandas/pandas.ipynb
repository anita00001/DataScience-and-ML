# Pandas Data Structures
- Series - One dimensional labelled array
- DataFrame - two dimensional table similar to a spreadsheet

```python
import pandas as pd
import numpy as np
```

### Series: Key features
- labelled -  index
- can store different data types

### Creating Series

```python
data = [3, 3.5, "anita"]
series = pd.Series(data)
print(series)
```

```python
pd.Series(data = [3, 7.7, "anita", 3, 7], index=['a', 'b', 'c', 'd', 'e'])
```

```python
# Write 10 names of students with custom index
pd.Series(data = ["Anita", "Krishna", "Pradip", "Sunita", "Anup", "Suman", "Desmond", "Erum", "Siddhartha", "Kuber"],
          index = ["a", "b", "c", "d", "e", "f", "g", "h", "i", "j"])
```

### DataFrames: Key Features
- Multiple columns
- can handle missing values
- supports indexing
### Creating dictionary from Series

```python
data = {
    "Name": ['Alice', 'Anita', 'Bob'],
    'Age': [25, 30, 34]
}
df = pd.DataFrame(data)
print(df)
```
