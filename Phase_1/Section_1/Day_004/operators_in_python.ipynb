# Operators
- perform operation on variables and values
- backbone of the computaion in python

## Operators in python
1. Arithmetic Operators
2. Comparison Operators (Relational operator): greater than, less than, equal to, greater than or equal to, ...
3. Logical Operators
4. Bitwise Operators (and, or, not, xor, leftwise)
5. Assignment Operators
6. Membership Operators: not in, in
7. Identity Operators: compare memory location: is, is not

```python
# Arithmetic Operator
a = 10
b = 3
print (a + b)
print (a-b)
print(a*b)
print(a**b)
print(a/b)
```

```python
# Comparison (Relational) Operator
x = 5
y = 10
print(x < y)
print(x > y)
print(x == y)
print(x != y)
```

```python
# Logical Operation: and, or, not
x = 10
y = 5
print(x>5 and y<6)
print(x<5 or y<6)
print(x<5 or y>6)
```

```python
# Bitwise operator: bit
a = bin(3)
b = bin(8)
print(a)
print(b)
```

```python
# Bitwise operator
a = 5
b = 3
print(a & b)
print(a | b)
print(a << 1)
print(b >> 1)
```
