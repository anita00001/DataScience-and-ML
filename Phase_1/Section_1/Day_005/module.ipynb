# Python Modules
- It is a python file that contains codes (functions, classes, variables) that are reused in other files.
- Module help organizing codes logically

Types of modules:
- in-built modules
- custom module

### In-built modules
- Import and use a module: we use the 'import' keyword

Examples of in-built modules:
- math module (https://docs.python.org/3/library/math.html)


```python
import math

print(math.sqrt(4))
```
