### Keyword arguments

### variable-length arguments (*args and **kwargs)

```python
# *args
def add_nums(*args):
    return sum(args)
```

```python
add_nums(1,2,3,4,5)
```

```python
add_nums(1,2,3)
```

```python
# keywords arguments **kwargs
## takes arguments like dictionary,  it has key and word(value)
def display_info(**kwargs):
    for key, value in kwargs.items():
        print(f"{key} : {value}")
```

```python
display_info(name = "Alice", age = 23, city = "Newyork")
```

```python
display_info(name = "Alice", age = 23, city = "Newyork", gender = "male")
```

### Scope of a variable
- local variable: found inside a function
- global variable: can be accessed outside a function

```python
def example():
    x = 10 # x as local variable
    print(x)

example() # x as local variable accessible inside a function only
```

```python
# x # x as local variable not accessible outside a function
```

```python
# global variable
y = 20 # y as global variable

def example2():
    print(y)
```

```python
y # y as global variable accessible outside a function
```

```python
example2() # y as global variable accessible inside a function
```

### Lambda Function (Anonymous Function)
- one line function
- 'lambda arguments: expression'
lambda = keyword
argument = argument
expression = just like mathematical expression, can be any

```python
square_num = lambda x: x*x
print(square_num(4))
```

### Function Documentation
- DocStrings to document the function

```python
def greeting(name):
    # DocStrings
    """This function prints a greetinng message.

    It takes the name of the person as argument and print greet message with their name.
    
    """ 
    print("Hello", name)
```

```python
# Print the documentation of the function
print(greeting.__doc__)
```
