## Set Operation
1. Union - all unique items from setA and setB
2. Intersection - common items in setA and setB
3. Set Difference : 
setA - setB : all items of setA that are not in setB
setB - setA : all items of setB that are not in setA
4. Symmetric Difference - Different items in setA and setB (not of intersection of sets)


```python
set1 = {1,2,3}
set2= {3,4,5}
```

```python
# set union
print (set1 | set2)
```

```python
# set intersection
print (set1 & set2)
```

```python
# set difference
print (set1 - set2)
```

```python
# symmetrical difference
print (set1 ^ set2)
```

# Dictionaries
- collection of elements
- unordered
- has key-value pair
- has unique keys
- mutable

```python
# create a dictionary
my_dict = {"name": "Alice", "age":25}
# name, age = key
# Alice, 25 = value
```

```python
my_dict
```

```python
# Accessing values with key
my_dict["name"]
```

```python
# modify the value
my_dict["name"] = "Bob"
```

```python
my_dict
```

```python
# Add new key-value pairs
my_dict["city"] = "Kathmandu"
my_dict
```

```python
# remove key-value pair
# delete method
del my_dict["age"]
my_dict
```
