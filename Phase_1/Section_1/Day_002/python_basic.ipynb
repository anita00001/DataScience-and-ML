### Variables in Python
- Naming Variables
1. Variables must start with a letter or and underscore (_)
2. Variables names cannot start with a number
3. Variable names can only contain letters, number and underscore 
4. Variables are case sensitive (Age and age are different variables)
5. You cannot use reserved words (if, while, class...)

```python
f_name = "anita"
l_name = "sharma"
age = 20
marks = 89.5
result = True
result = "Pass" if True else "Fail"
```

```python
# f-string
f_string = f"She is {f_name} {l_name}, age {age}. She has obtained {marks}%. She {result} the exam"
f_string
```

```python
# input
name = input("Enter your name: ")
print(f"Hello, {name}")
```

- Ordered : have index
- Mutable (Changable) : Elements can be added, removed or modified
- Dynamic : Grow or shrink as needed
- Enclosed in large ([]) bracket, separated by comma(,)

```python
# List of integer
list_num = [1, 2, 3, 5, 7, 11] # List of prime numbers
list_num
```

```python
type(list_num)
```

```python
# List of float
list_dec = [0.1, 0.2, 0.3, 0.5, 0.7, 0.11] # List of prime_numbers/10
list_dec
```

```python
type(list_dec)
```

```python
# List of string
list_str = ["Apple", "Banana", "Cherry", "Orange"] # List of fruits
list_str
```

```python
# Access a element on list
list_fruit = ['apple', 'banana', 'cherry', 'grape', 'orange']
list_1 = list_fruit[0] # apple
list_1
```

```python
type(list_1)
```

```python
# Access last element on the list : use pop method
list_fruit = ['apple', 'banana', 'cherry', 'grape', 'orange']
list_pop = list_fruit.pop() # orange
list_pop
```

```python
type(list_pop)
```

```python
# Access elements on list: list_name[start:end:step] --> end means one index ahead of end
# start=1, end=3, step=1
list_fruit = ['apple', 'banana', 'cherry', 'grape', 'orange']
list_2 = list_fruit[1:3] # banana, cherry
list_2
```

```python
type(list_2)
```

```python
# Access alternate elements on list
# start=0, end=last, step=2
list_fruit = ['apple', 'banana', 'cherry', 'grape', 'orange']
list_3 = list_fruit[::2] # apple, cherry, orange
list_3
```

```python
# Access alternate elements on list
# start=0, end=last, step=3
list_fruit = ['apple', 'banana', 'cherry', 'grape', 'orange']
list_4 = list_fruit[::3] # apple, grape
list_4
```

```python
# Access specific elements on list : banana, grape
# start=1, end=4, step=2
list_fruit = ['apple', 'banana', 'cherry', 'grape', 'orange']
list_5 = list_fruit[1:4:2] # banana, grape
list_5
```

```python
# reverse list
# start=0, end=last, step=-1 : since step is -1, it search from index:last to index:0
list_int = [1,2,3,4,5]
reversed_list = list_int[::-1]
reversed_list
```

```python
# Concat two lists
list_odd = [1,3,5]
list_even = [2,4]
list_int = list_odd + list_even
list_int
```

```python
# Sort list
list_int.sort()
```

```python
list_int
```
