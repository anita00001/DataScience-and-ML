### Geographical Information System

- According to ESRI, G.I.S is a
technology used to collect all types
of geographical data. This data is
used to represent, analyze, and
visualize spatial and geographical
information

#### Types of GIS Data
1. Vector data: Points, Lines, Polygons
2. Raster data: Picture, Images (Pixels, Grid: each cell contains a value
representing a feature(e.g.
temperature, elevation, etc.))

#### GIS Data Format
- .shp (Shapefile: vector format containing spatial and attribute data)
- .geojson (GeoJSON: json-based format for encoding geographic data)
- .kml (KML: used in Google Earth for geographic visualization)
- .geotiff, .jpg, .png (Raster Format: used for images, satellite data, digital elevation models)
- .csv (Comma Separated Value)
- .kmz (KMZ: used in google Earth as a storage format)
- .dwg (DWG: to store 2D or 3D vector graphics)

#### GIS Data Sources:
1. Primary Data Sources: collected using Remote Sensing, Survey Data, GPS Devices
2. Secondary Data Sources: reuse from earlier sources: Public GIS Databases (OpenStreetMap, Government Portals, etc), Scanned paper maps

#### GIS Coordinate Systems and Projections
- To define location
##### Types:
- GCS (Geographic Coordinate
System): Uses latitude and
longitude
- PCS (Projected Coordinate System): Transforms the Earth 3D shape onto
a 2D map

#### GIS Applications
- Disaster Management
- Air Pollution and Control
- Natural Resource Management
- Waste Water and Management
- Coal Mine Fire Management
- Forest Fire Management
- Oil Spill and remedial Measures

-------------------------------------
* Vector Data: .shp, .geojson
* Raster Data: .geotiff, .jpg
-------------------------------------

#### Working with Vector Data

```python
import geopandas as gpd
import matplotlib.pyplot as plt
```

```python
# Load a shapefile data
path = r"gis-data/ServiceDistricts.shp"
# Reading a shapefile data
gdf = gpd.read_file(path)

# .shx has index while .shp is the original file with the geolocations stored.
```

```python
gdf.head()
```

```python
# check the data types
gdf.dtypes
```

```python
# plot the shapefile data
# gdf.plot()
gdf.plot(figsize=(10,6), edgecolor = 'b', cmap="viridis")
plt.title("Simple shapefile map")
plt.show()
```

```python
# Loading a GeoJSON data
path = r"gis-data/school-district.geojson"
geo_df = gpd.read_file(path)
```

```python
geo_df.head()
```

```python
geo_df.plot(figsize=(10,6), cmap="viridis", edgecolor='black')
plt.title("GeoJSON Map Illustration")
plt.show()
```

#### CRS (Coordinate Referencing System)
- The CRS defines how the geometric data is mapped onto the Earth's surface, and it can be a Proected Coordinate System (eg:UMT) or a Geographic Coordinate System (eg: WGS84)

```python
print(geo_df.crs)
```

- ✍EPSG: European Petroleum Survey Group code format

```python
# Reproecting to a different CRS
geo_dft = geo_df.to_crs(epsg=3857)
```

```python
geo_dft.crs
```

### Working with Raster Files

```python
import rasterio
```

```python
# Open the raster file
raster_path = r"gis-data/sample.tif"
# load raster data
src = rasterio.open(raster_path)
```

```python
# Read the raster data (first band)
raster_bd1 = src.read(1)  # Read the first band: indexed-1
# band: images are made up of different bands
raster_bd1
```

```python
# Get spatial extent (bounds) of the raster
bounds = src.bounds
print(f'Raster Bounds: {bounds}')
```

```python
# plot the raster data
plt.figure(figsize=(8,8))
plt.imshow(raster_bd1, cmap='viridis') # gray, plasma, coolwarm
plt.colorbar(label='Raster Value')
plt.title('Raster Data Visualization')
plt.show()
```

```python
raster_bd2 = src.read(2)
```

```python
# plot the raster data
plt.figure(figsize=(8,8))
plt.imshow(raster_bd2, cmap='plasma')
plt.colorbar(label='Raster Value')
plt.title('Raster Data Visualization')
plt.show()
```
