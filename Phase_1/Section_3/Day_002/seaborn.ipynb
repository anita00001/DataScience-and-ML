## Example comparison

```python
x = [1,2,3,4,5]
y = [10,20,25,30,35]
```

```python
# matplotlib
plt.scatter(x,y, color='b')
plt.xlabel("x-values")
plt.ylabel('y-values')
plt.title("matplotlib scatter plot")
plt.show()
```

```python
# Seaborn
sns.scatterplot(x=x, y=y, color='r')
plt.title("Seaborn Scatter plot")
plt.show()
```

```python
print(sns.get_dataset_names())
```

```python
# load the datasets in seaborn (eg: tips dataset)
df = sns.load_dataset('tips')
df.head() # view first 5 rows
```

#### Basic plots
   - Scatter plot
   - line plots
   - Histogram
   - Bar Plots
   - Box Plots

```python
# Scatter plot: Can only be produce for numerical data 
sns.scatterplot(x="total_bill", y="tip", color='g', data=df)
plt.title("Scatter plot of total bill vs Tip")
plt.show()
```

```python
# Line plot
sns.lineplot(x="size", y="total_bill", data=df)
plt.show()
```

```python
# Histogram -  Used for ploting distribution of data
sns.histplot(df['total_bill'], bins=20) # bins: number of rectangle
plt.title("Histogram of Total bill")
plt.show()
```

```python
sns.histplot(df['total_bill'], bins=20, kde=True) # kde: kannel density estimation :Produce follow-line
plt.title("Histogram of Total bill")
plt.show()
```

```python
# Bar plot - Categorical data
sns.barplot(x= 'day', y='total_bill', data=df)
plt.title("Barplot of the total bill by day")
plt.show()
```

```python
# Boxplot - Used to detect outliers
# Identify: 1. outlier, 2. minimum value, 3. maximum value, 4. median (Q2(50 percentile of data)), 5. quartile: Q1 (25 percentile of data) and Q3 (75 percentile of data)
sns.boxplot(x='day', y='total_bill', data=df)
plt.title("Box plot of total bill by day")
plt.show()
```

#### Advanced Visualization
   - Violin Plot
   - pairplot
   - Heatmaps
   - KDE plots

```python
# Violin Plot
# comibination of boxplot and density plot
# # Identify: 1. outlier, 2. minimum value, 3. maximum value, 4. median (Q2(50 percentile of data)), 5. quartile: Q1 (25 percentile of data) and Q3 (75 percentile of data)
# Buldges out at maximum density data
sns.violinplot(x='day', y='total_bill', data=df)
plt.show()
```

```python
# Pairplots - Scatter matrix
sns.pairplot(df, hue='sex')
plt.show()
```

```python
df2 = df[['total_bill', 'tip', 'size']]
```

```python
# Heatmap - used for visualization of correlation matrix
# correlation = (change in y)/(change in x)
corr = df2.corr()
sns.heatmap(corr, annot=True, cmap='coolwarm', linewidth=0.5)
plt.show()
```

```python
# KDE plot - Kernel Density Estimation
sns.kdeplot(df['total_bill'], fill=True, color='r')
plt.show()
```

### Customizing Seaborn Plot

```python
# Changing colors
# https://seaborn.pydata.org/generated/seaborn.set_palette.html
sns.set_palette('pastel')
sns.boxplot(x='day', y='total_bill', data=df)
plt.show()
```

```python
# Adding title and labels
sns.scatterplot(x='total_bill', y='tip', data=df, color='g')
plt.xlabel("Total Bill ($)")
plt.ylabel("Tip ($)")
plt.title("Scatter Plot")
plt.show()
```

```python
## Adjusting styles
sns.set_style("darkgrid") # white, dark, darkgrid, whitegrid, ticks
sns.histplot(df['total_bill'], bins=20)
plt.show()
```

### Combining multiple plots

```python
fig, axes = plt.subplots(1,2, figsize=(12,5))

sns.boxplot(x='day', y='total_bill', data=df, ax=axes[0])
axes[0].set_title("Box Plot")

sns.histplot(df['total_bill'], bins=10, kde=True, ax=axes[1])
axes[1].set_title("Histogram")

plt.show()
```

### Save plots

```python
sns.boxplot(x='day', y='total_bill', data=df)
plt.savefig("boxplot.png", dpi=300) # saving the box plot as png file, dpi = resolution
```
